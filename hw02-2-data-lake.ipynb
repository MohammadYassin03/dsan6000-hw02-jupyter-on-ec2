```python
import os
import boto3

NETID = "my677"
BUCKET = f"dsan6000-{NETID}"
REGION = "us-east-1"

# Default credentials (the same ones the aws CLI uses on your EC2 instance)
s3 = boto3.client("s3", region_name=REGION)
```

```python
s3.create_bucket(Bucket=BUCKET)
```

```python
for fname in sorted(os.listdir("data")):
    if fname.endswith(".parquet"):
        s3.upload_file(
            os.path.join("data", fname),
            BUCKET,
            f"wikipedia-hourly/{fname}",
        )
print("uploaded")
```

```python
resp = s3.list_objects_v2(Bucket=BUCKET, Prefix="wikipedia-hourly/")
print(resp["KeyCount"])
[o["Key"] for o in resp["Contents"]][:3]
```
